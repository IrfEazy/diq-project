```python
DESCRIPTION = 'Completeness - MCAR with different low percentages imputing the standard deviation'
RESULTS_PER_ALGORITHM = []

for algorithm in REGRESSION_ALGORITHMS:  # FIRST CYCLE ON THE ALGORITHMS
    RESULTS_ALGORITHM = []

    for i in range(len(X_AXIS)):
        # DATA COLLECTION - creating 10 datasets with an increase in the samples
        X, y = make_dataset_for_regression(
            n_samples=1000, n_features=3, n_informative=3, n_targets=1, bias=0.0,
            effective_rank=None, tail_strength=0.5, noise=0.0, seed=2023
        )

        X = pd.DataFrame(X)

        X = pollute_data_mcar(X, percent_incomplete=X_AXIS[i])

        # every column's NaN are filled with that column's standard deviation
        X_CLEAN = X.fillna(X.std())

        # DATA ANALYSIS
        RESULTS_ALGORITHM.append(regression(X_CLEAN, y, algorithm, SEED))

    RESULTS_PER_ALGORITHM.append(RESULTS_ALGORITHM)

plot_results(
    x_axis_values=X_AXIS, x_label=X_LABEL, results=RESULTS_PER_ALGORITHM, title=DESCRIPTION,
    algorithms=REGRESSION_ALGORITHMS, cleaned_data=True
)
```

```python
DESCRIPTION = 'Completeness - different MCAR low percentages imputing with KNN'

RESULTS_PER_ALGORITHM = []

IMPUTER = KNNImputer(n_neighbors=5)

for algorithm in REGRESSION_ALGORITHMS:  # FIRST CYCLE ON THE ALGORITHMS
    RESULTS_ALGORITHM = []

    for i in range(len(X_AXIS)):
        # DATA COLLECTION - creating 10 datasets with an increase in the samples
        X, y = make_dataset_for_regression(
            n_samples=1000, n_features=3, n_informative=3, n_targets=1, bias=0.0,
            effective_rank=None, tail_strength=0.5, noise=0.0, seed=2023
        )

        X = pd.DataFrame(X)

        X = pollute_data_mcar(X, percent_incomplete=X_AXIS[i])

        # KNN imputation of the whole frame, keeping columns and index
        X_CLEAN = pd.DataFrame(IMPUTER.fit_transform(X), columns=X.columns, index=X.index)

        # DATA ANALYSIS
        RESULTS_ALGORITHM.append(regression(X_CLEAN, y, algorithm, SEED))

    RESULTS_PER_ALGORITHM.append(RESULTS_ALGORITHM)

plot_results(
    x_axis_values=X_AXIS, x_label=X_LABEL, results=RESULTS_PER_ALGORITHM, title=DESCRIPTION,
    algorithms=REGRESSION_ALGORITHMS, cleaned_data=True
)
```

```python
DESCRIPTION = 'Completeness - different MCAR low percentages imputing with Linear Regression'
RESULTS_PER_ALGORITHM = []

for algorithm in REGRESSION_ALGORITHMS:  # FIRST CYCLE ON THE ALGORITHMS
    RESULTS_ALGORITHM = []

    for i in range(len(X_AXIS)):
        # DATA COLLECTION - creating 10 datasets with an increase in the samples
        X, y = make_dataset_for_regression(
            n_samples=1000, n_features=3, n_informative=3, n_targets=1, bias=0.0,
            effective_rank=None, tail_strength=0.5, noise=0.0, seed=2023
        )

        X = pd.DataFrame(X)

        X = pollute_data_mcar(X, percent_incomplete=X_AXIS[i])

        # Per-column regression imputation on the whole frame
        X_CLEAN = X.copy()
        for column in X.columns:
            missing = X[column].isna()
            if not missing.any():
                continue
            # predictors = the other columns, mean-filled
            predictors = X.drop(columns=[column])
            predictors = predictors.fillna(predictors.mean())
            # fit on the rows where the target column is observed, predict the missing ones
            model = linear_model.LinearRegression()
            model.fit(predictors[~missing], X.loc[~missing, column])
            X_CLEAN.loc[missing, column] = model.predict(predictors[missing])

        # DATA ANALYSIS
        RESULTS_ALGORITHM.append(regression(X_CLEAN, y, algorithm, SEED))

    RESULTS_PER_ALGORITHM.append(RESULTS_ALGORITHM)

plot_results(
    x_axis_values=X_AXIS, x_label=X_LABEL, results=RESULTS_PER_ALGORITHM, title=DESCRIPTION,
    algorithms=REGRESSION_ALGORITHMS, cleaned_data=True
)
```

Dataset 10 -> 3 features like cell 5. Unused `KNNImputer` instance removed.

- Cells 22-23 (exp. 5, MAR), 25-26 (exp. 6, MNAR), 28-35 (exp. 7-9), 37-38 (exp. 10): same calls, but now backed by the v2 functions (quantile-based MAR/MNAR that actually create NaN, NaN instead of a constant for the most/less important features, last three features of the ranking for the less important ones, see `library/v2/CHANGES.md`).
